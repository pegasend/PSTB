```python
# 1-list(tuples) => dico([tuple()])=dico
# dict(tuple(k,v) )=dico{k,v}
l_tuple=[(1,'red'),(2,'blue'),(3,'green')]
dico=dict(l_tuple)
dico
```

```python
#2-2xList => dico=dict(zip(l1,l2) )
l_cle=[1,2,3,4]
l_value=['a','b','c','d']
dict(zip(l_cle,l_value) )
```

```python
#3-2xList de longueur différentes => dico=dict(zip_longest(l1,l2) )
from itertools import zip_longest
l_cle=[1,2,3,4,5,6,7]
dict(zip_longest(l_cle, l_value, fillvalue='x') )# rempli les None par 'x'
#{1: 'a', 2: 'b', 3: 'c', 4: 'd', 5: None, 6: None, 7: None}
```

```python
# 3- merge dico=dico1+dico2
d1={1:'a',2:'b'} # d1=dict([(1,'a'),(2,'b')])
d2={3:'c',4:'d'}
merge_dico={}
for dico in [d1,d2]:
    merge_dico.update(dico)
print(merge_dico)
```

```python
# exo1
def diff(a,b):
    return a-b
diff(2,2),diff(0,2)
```

```python
def print_day(j): # renvoi le jour du n° de la semaine
    jour={2:"lundi",3:"mardi",4:'mercredi',5:'jeudi',6:'vendredi',7:'samedi',1:'dimanche'}
    if j in jour.keys():
        return jour[j]
    else:
        return None
print_day(4), print_day(41) 
```

```python
def last_element(l): # retourne le dernier élt
    if len(l)==0:
        return None
    return l[-1]
assert last_element([1,2,3,4])==4
assert last_element([])==None
```

```python
def compare(x,y):
    if x>y:return "first is greater"
    elif x<y:return "second is greater"
    else:return "equal"
assert compare(1,1)=="equal"
assert compare(1,2)=="second is greater"
assert compare(2,1)=="first is greater"
```

```python
def letter_count(w,l):
    w=w.lower()
    l=l.lower()
    if l in w:
        return w.count(l)
    else:
        return 0
assert letter_count("amazing",'A')==2
```

```python
from collections import Counter
def letter_count(s):
    return dict(Counter(s) )
assert letter_count("hello")=={'h':1, 'e': 1, 'l': 2, 'o':1}
assert letter_count("person")=={'p':1, 'e': 1, 'r': 1, 's':1, 'o':1, 'n':1}
```

```python
def manip_list(l,f,pos='fin', n=0):
    if f=='remove':
        if pos=='fin':
            return l.pop(-1)
        elif pos=='debut':
            return l.pop(0)
    elif f=='add':
        if pos=='debut':
            return [n]+l
        elif pos=='end':
            return l+[n]
    return
l=[1,2,3]
assert manip_list(l.copy(), "remove", "fin")==3 # le dernier
assert manip_list(l.copy(), "remove", "debut")==1 # le premier
assert manip_list(l.copy(), "add", "debut", 20)==[20,1,2,3] # ajouter la valeur au debut et renvois la liste
assert manip_list(l.copy(), "add", "end", 30)==[1,2,3,30] # ajouter la valeur à la fin et renvois la liste
```

```python
def palindrome(s):# se lit de la même manière à l'envers ou à l'endroit
    if s.lower().replace(" ","")==s[::-1].lower().replace(" ",""):
        return True
    else:
        return False
assert palindrome('testing')==False
assert palindrome('tacocat')==True
assert palindrome('hannah')==True
assert palindrome('robert')==False
assert palindrome('a man a plan a canal Panama')==True
```

```python
def freq(liste, item):
    return liste.count(item)
assert freq([1,2,3,4,4,4], 4)==3
assert freq([True, False, True, True], False)==1
```

```python
def flip_case(s,l):# inverse la casse de toutes les occurrences
    return "".join([c.upper() if (l==c.lower() and c.islower() ) else c.lower() if (l==c.lower() and c.isupper() ) else c for c in s])

flip_case("Hardy har har", "h")
```

```python
import math
def multipl_paire_nb(l):
    return math.prod([pair for pair in l if pair%2==0]) # check si paire, puis multiplie ts les elts de la liste
assert multipl_paire_nb([2,3,4,5,6])==48
```

```python
from collections import Counter
# récuperer la clé d'un dico selon sa value
def mode(l): # la fréquence la + importante d'une liste
    dico_freq=dict(Counter(l) ) # {2: 2, 4: 7, 1: 1, 3: 2, 5: 1, 6: 2, 7: 1} occurrence pour chaque keys
    return next(k for k,v in dico_freq.items() if v==max(dico_freq.values() ) ) # filtre sur la freq max et donne sa clé pour dico_freq

assert mode([2,4,1,2,3,3,4,4,5,4,4,6,4,6,7,4])==4
```

```python
def first_majuscule(s): # renvoi la 1ère lettre en majuscule
    return s.title()
def first_majuscule2(s):
    return s[0].upper()+s[1:]

assert first_majuscule("tim")=="Tim"
assert first_majuscule2("matt")=="Matt"
```

```python
def compact_boolean(l):# retourne les valeur True
    return [elt for elt in l if bool(elt)]
assert compact_boolean([0,1,2,"",[], False, {}, None, "All done"])==[1,2, "All done"]
```

```python
def is_even(num): # check si c'est paire, renvoie True sinon False
    return num % 2 == 0

def partition(l, func_paire): # appel is_even() et ajout dans une liste les True et autre liste pour les False, puis ajout des 2 listes dans une liste
    l_paire=[]
    l_impaire=[]
    for n in l:
        if func_paire(n):# split les paires
            l_paire.append(n)
        else: # split les impaires
            l_impaire.append(n)

    return [l_paire, l_impaire] # 2 listes dans une liste

assert partition([1,2,3,4], is_even)==[[2,4],[1,3]]
```

```python
def intersection(l1,l2): # 2 listes et renvoit l'intersection commune
    return [i for i in l1 if i in l2]

def intersection2(l1,l2):
    return list(set(l1).intersection(l2) )# intersection de collection ou "set"

assert intersection([1,2,3], [2,3,4])==[2,3]
assert intersection2([1,2,3], [2,3,4]) # [2,3]
```

```python
# https://stackoverflow.com/questions/53327829/why-do-these-two-approaches-print-different-results
def add(a,b):
    return a + b

def once(func): # vous devrez définir une nouvelle fonction à l'intérieur de votre fonction once et la renvoyer
    fonction = [func]
    def func_interne(*args):
        print("id=",id(fonction) )#  propriétés à votre fonction interne pour vérifier si elle a déjà été exécutée

        return fonction.pop()(*args) if len(fonction) else None # nv fonction invocable une seule fois
    
    return func_interne # renvoyer la fonction interne

one_addition = once(add)

assert one_addition(2,2)==4
assert one_addition(2,2)==None
assert one_addition(12,200)==None
```

```python
#Super bonus: https://cosmiccoding.com.au/tutorials/handy_python_decorators/
def run_once(fn):
    def wrapper(*args, **kwargs):
        if not wrapper.has_run:
            wrapper.has_run = True
            return fn(*args, **kwargs)
        else:
            print("NO! NOT ALLOWED!")
    wrapper.has_run = False
    return wrapper

@run_once
def add(a, b):
    return a + b

# print(add(6, 7) )
# print(add(7, 8) )
assert add(2,2)==4
assert add(2,2)==None
assert add(12,20)==None
```
